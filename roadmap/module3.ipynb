## Task 1
Show that the RBF kernel $k(x,x') = \exp(-\|x-x'\|^2 / 2\ell^2)$
   corresponds to an infinite-dimensional feature map.

### From a feature map to a kernel

With $f(x) = \phi(x)^\top \beta$ and $\beta \sim \mathcal{N}(0, I)$ the core equation gives

$$k(x, x') = \phi(x)^\top \phi(x') = \sum_{n} \phi_n(x)\,\phi_n(x').$$

WLOG $\ell = 1$, so we can ignore it.

So we have to find features $\phi_n$, infinitely many of them, whose inner product is the RBF kernel. For $\Sigma_0 \neq I$ one absorbs $\Sigma_0^{1/2}$ into $\phi$ (R&W §2.1.2), so $\Sigma_0 = I$ loses nothing.

**Taylor expansion (1-D).** Expand the square in the exponent, $-\dfrac{1}{2} (x-x')^2 = -\dfrac{1}{2} x^2 - \dfrac{1}{2} x'^2 + xx'$. The cross term is the only one that couples $x$ and $x'$, and it is the one that gets a power series:

$$k(x,x') = \exp\!\left(-\frac{1}{2} x^2\right) \exp\!\left(-\frac{1}{2} x'^2\right) \exp(xx')
= \sum_{n=0}^{\infty} \underbrace{\exp\!\left(-\frac{1}{2} x^2\right) \frac{1}{\sqrt{n!}} x^n}_{\phi_n(x)} \underbrace{\exp\!\left(-\frac{1}{2} x'^2\right) \frac{1}{\sqrt{n!}} x'^n}_{\phi_n(x')}$$

The series of $\exp(z)$ converges for every $z$, so the sum converges for all $x, x'$. The feature map $\phi(x) = (\phi_0(x), \phi_1(x), \dots)$ is a Gaussian envelope times all monomials.

$$\phi_n(x) = \exp\!\left(-\frac{1}{2} x^2\right) \frac{1}{\sqrt{n!}} x^n$$
$$\phi(x) = \exp\!\left(-\frac{1}{2} x^2\right) \begin{pmatrix} 1 \\ \,x \\ \frac{1}{\sqrt{2}}\,x^2 \\ \vdots \end{pmatrix}$$

  
**More dimensions.** $\exp(x\cdot x') = \sum_n \dfrac{(x\cdot x')^n}{n!}$ and by the multinomial theorem $\dfrac{(x\cdot x')^n}{n!} = \sum_{|\alpha| = n} \dfrac{x^\alpha x'^\alpha}{\alpha!}$, a finite sum for each $n$. The features are $\phi_\alpha(x) = \exp\!\left(-\dfrac{1}{2} \|x\|^2\right) \dfrac{1}{\sqrt{\alpha!}} x^\alpha$ over all multi-indices $\alpha$, so the same conclusion holds.

  
**Why infinitely many are really needed.** If $k(x,x') = \phi(x)^\top\phi(x')$ with $\phi \in \mathbb{R}^N$, every Gram matrix $K = \Phi\Phi^\top$ has rank at most $N$. The RBF Gram matrix on $n$ distinct points is positive definite (rank $n$) for every $n$, which follows from Bochner's theorem (R&W Thm. 4.1) since the spectral density $S(s) \propto \exp(-2\pi^2 s^2)$ is positive everywhere. So $N \ge n$ for every $n$, i.e. $N = \infty$. This is not a limitation of the Taylor construction: *no* finite feature map exists.

**Answer.** Writing $\exp\!\left(-\dfrac{1}{2} (x-x')^2\right)$ as $\exp\!\left(-\dfrac{1}{2} x^2\right) \exp\!\left(-\dfrac{1}{2} x'^2\right) \exp(xx')$ and Taylor-expanding the last factor gives $k(x,x') = \sum_{n=0}^{\infty} \phi_n(x)\phi_n(x')$ with $\phi_n(x) = \exp\!\left(-\dfrac{1}{2} x^2\right) \dfrac{1}{\sqrt{n!}} x^n$. So the RBF kernel is the covariance of a Bayesian linear model with $\Sigma_0 = I$ and infinitely many basis functions, and no finite number can do, since a finite feature map would limit the rank of every Gram matrix.

## Task 2
Why does moving to function space remove the need to ever
   specify or invert a weight-space covariance matrix?

### Weight space vs function space

Let $\Phi \in \mathbb{R}^{n\times N}, \Phi_\ast \in \mathbb{R}^{m\times N}$ be the matrices of features of the training and test inputs (one row per point), and  
$f = \Phi\beta$, $\beta \sim \mathcal{N}(0, \Sigma_0)$, $y = f + \varepsilon$, $\varepsilon \sim \mathcal{N}(0, \sigma_n^2 I)$.  

<br>  

**Weight space** (R&W eq. 2.11) needs the $N \times N$ matrix $A$, which contains $\Sigma_0^{-1}$:

$$A = \frac{\Phi^\top\Phi}{\sigma_n^2} + \Sigma_0^{-1}, \qquad
f_\ast \mid y \sim \mathcal{N}\!\left(\tfrac{1}{\sigma_n^2}\Phi_\ast A^{-1}\Phi^\top y,\;\; \Phi_\ast A^{-1}\Phi_\ast^\top\right)$$   

<br>  

**Function space** (R&W eq. 2.12) needs only the $n \times n$ matrix $K + \sigma_n^2 I$, where $K = \Phi\Sigma_0\Phi^\top$, $K_\ast = \Phi_\ast\Sigma_0\Phi^\top$ and $K_{\ast\ast} = \Phi_\ast\Sigma_0\Phi_\ast^\top$ are evaluations of $k(x, x') = \phi(x)^\top\Sigma_0\phi(x')$:

<br>  

$$f_\ast \mid y \sim \mathcal{N}\!\left(K_\ast (K + \sigma_n^2 I)^{-1} y,\;\; K_{\ast\ast} - K_\ast (K + \sigma_n^2 I)^{-1} K_\ast^\top\right)$$

**They are the same.** Since $K_\ast = \Phi_\ast\Sigma_0\Phi^\top$, the two means agree if

$$\tfrac{1}{\sigma_n^2}\,A^{-1}\Phi^\top \;=\; \Sigma_0\Phi^\top\,(K + \sigma_n^2 I)^{-1}. \tag{*}$$

*Step 1: multiply $A$ by $\Sigma_0\Phi^\top$* (an $N \times n$ matrix). The factor $\Sigma_0^{-1}$ cancels, then $\Phi^\top$ is pulled out to the left and $\Phi\Sigma_0\Phi^\top = K$:

$$A\,\Sigma_0\Phi^\top
= \frac{\Phi^\top\Phi\,\Sigma_0\Phi^\top}{\sigma_n^2} + \Sigma_0^{-1}\Sigma_0\Phi^\top
= \frac{\Phi^\top\left(\Phi\Sigma_0\Phi^\top + \sigma_n^2 I\right)}{\sigma_n^2}
= \frac{1}{\sigma_n^2}\,\Phi^\top\,(K + \sigma_n^2 I).$$

*Step 2: move the inverses.* Multiply by $A^{-1}$ from the left and by $(K + \sigma_n^2 I)^{-1}$ from the right (both matrices are positive definite, so both inverses exist). This gives $(\ast)$.

*Step 3: insert $(\ast)$ into the weight-space mean.* With $K_\ast = \Phi_\ast\Sigma_0\Phi^\top$:

$$\tfrac{1}{\sigma_n^2}\,\Phi_\ast A^{-1}\Phi^\top y
= \Phi_\ast\left[\Sigma_0\Phi^\top (K + \sigma_n^2 I)^{-1}\right] y
= K_\ast\,(K + \sigma_n^2 I)^{-1} y.$$

The left side inverts the $N \times N$ matrix $A$, the right side the $n \times n$ matrix $K + \sigma_n^2 I$.

*Covariance.* The matrix inversion lemma gives $A^{-1} = \Sigma_0 - \Sigma_0\Phi^\top (K + \sigma_n^2 I)^{-1}\Phi\Sigma_0$. Multiplying by $\Phi_\ast$ from the left and $\Phi_\ast^\top$ from the right:

$$\Phi_\ast A^{-1}\Phi_\ast^\top = \Phi_\ast\Sigma_0\Phi_\ast^\top - \Phi_\ast\Sigma_0\Phi^\top (K + \sigma_n^2 I)^{-1}\Phi\Sigma_0\Phi_\ast^\top = K_{\ast\ast} - K_\ast (K + \sigma_n^2 I)^{-1} K_\ast^\top.$$

In the second form $\Sigma_0$ and $\phi$ occur only inside $k$, and $\Sigma_0^{-1}$ does not occur at all.

**Answer.** Because $\beta$ is integrated out analytically. The prior over the function values at any points $x_1, \dots, x_m$ is $f \sim \mathcal{N}(0, \Phi\Sigma_0\Phi^\top)$, and the whole of $\Phi\,\Sigma_0\,\Phi^\top$ is the kernel matrix $[k(x_i, x_j)]$. $\Sigma_0$ and $\phi$ never need to be known separately, only the number $k(x,x')$ for each pair of inputs. Prediction (eq. 2.12) then uses only

- the kernel matrices $K$, $K_\ast$, $K_{\ast\ast}$ and
- the inverse of $K + \sigma_n^2 I$, an $n \times n$ matrix, the covariance of the *observations*.

$\Sigma_0^{-1}$ is never formed, and the number of weights $N$ no longer enters.

This matters most for the RBF kernel. By Task 1 its feature map has infinitely many components, so $\Phi$ has infinitely many columns and the weight-space formulas (eq. 2.11) would need the $\infty \times \infty$ matrices $\Sigma_0$, $\Sigma_0^{-1}$ and $A$. They cannot even be written down. The function-space formulas need only the values $k(x_i, x_j)$ on the $n$ training and $m$ test points, which are available in closed form.

**Caveats**
- Function space does not remove matrix inversion. $K + \sigma_n^2 I$ is still inverted, which costs $O(n^3)$ and is the bottleneck that motivates Module 4 and eventually PFNs. What disappears is the weight-space covariance.
- Function space is the cheaper side when $N > n$ (infinite for RBF) and weight space when $N < n$ (e.g. a few polynomial features and many points).
